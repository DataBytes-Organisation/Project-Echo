### Event Segmentation Using Yamnet

```python
logging.basicConfig(level=logging.INFO)
warnings.filterwarnings("ignore")

# Load YAMNet
yamnet_model = yamnet.yamnet_frames_model(params)
yamnet_model.load_weights('yamnet.h5')
yamnet_classes = yamnet.class_names('yamnet_class_map.csv')

# Function to split a single audio file
def split_audio(input_file, output_dir, min_dur=1.2, max_dur=2, max_silence=0.5, energy_threshold=55):
    audio_regions = split(
        input_file,
        min_dur=min_dur,
        max_dur=max_dur,
        max_silence=max_silence,
        energy_threshold=energy_threshold,
        sw=2
    )
    for i, r in enumerate(audio_regions):
        print(f"Region {i}: {r.meta.start:.3f}s -- {r.meta.end:.3f}s")
        should_save = recognize_audio(r)
        if should_save:
            filename = os.path.join(output_dir, f"region_{r.meta.start:.3f}-{r.meta.end:.3f}.wav")
            r.save(filename)
            print(f"Region saved as: {filename}")

def recognize_audio(audio_region):
    waveform = audio_region.samples
    if waveform.ndim == 2 and waveform.shape[0] == 2:
        waveform = waveform[1, :]
    if waveform.ndim == 2:
        waveform = np.mean(waveform, axis=0)
    sample_width = 2
    waveform = waveform.astype(np.int16)
    waveform = waveform.astype(np.float32) / 32768.0
    prediction = np.mean(yamnet_model.predict(np.reshape(waveform, [1, -1]), steps=1)[0], axis=0)
    sorted_indices = np.argsort(prediction)[::-1]
    matching_classes = set(yamnet_classes[sorted_indices]).intersection(set(['Dog', 'Cat', 'Bird', 'Animal', 'Birdsong', 'Canidae', 'Feline', 'Livestock', 'Rodents, Mice', 'Wild animals', 'Pets', 'Frogs', 'Insect', 'Snake', 'Domestic animals, pets', 'crow']))
    is_condition_met = any([prediction[idx] > 0.2 for idx in sorted_indices if yamnet_classes[idx] in matching_classes])
    return is_condition_met

def process_batch(files_batch, folder_path, output_dir, catalog_number_mapping):
    for file in files_batch:
        logging.info(f"Starting processing for file: {file}")
        if file.endswith(".mp3") or file.endswith(".wav"):
            catalog_number = file.split('.')[0]
            if catalog_number in catalog_number_mapping:
                scientific_name = catalog_number_mapping[catalog_number]
                scientific_folder = os.path.join(output_dir, scientific_name)
                os.makedirs(scientific_folder, exist_ok=True)
                file_path = os.path.join(folder_path, file)
                split_audio(file_path, scientific_folder)

    logging.info(f"Batch processing completed, clearing memory...")
    tf.keras.backend.clear_session()
    gc.collect()

def main():
    folder_path = r"D:\AnimalSounds"
    output_dir =r"D:\Sorted_AnimalSounds"
    csv_file = r"D:\Event_Segmentation_YamNet\eBird data.csv"
    catalog_number_mapping = {}
    with open(csv_file, mode='r', encoding='utf-8') as csvfile:
        csv_reader = csv.reader(csvfile)
        header = next(csv_reader)
        for row in csv_reader:
            if len(row) >= 2:
                catalog_number = row[0].strip()
                scientific_name = row[3].strip()
                catalog_number_mapping[catalog_number] = scientific_name

    all_files = [file for root, dirs, files in os.walk(folder_path) for file in files if file.endswith((".mp3", ".wav"))]
    batch_size = 100
    total_files = len(all_files)
    num_batches = (total_files // batch_size) + (1 if total_files % batch_size != 0 else 0)

    for i in range(num_batches):
        start_idx = i * batch_size
        end_idx = start_idx + batch_size
        current_batch = all_files[start_idx:end_idx]
        logging.info(f"Processing batch {i + 1} of {num_batches}...")
        process_batch(current_batch, folder_path, output_dir, catalog_number_mapping)

    logging.info(f"Processing completed. Total files processed: {total_files}")

if __name__ == "__main__":
    main()
```
