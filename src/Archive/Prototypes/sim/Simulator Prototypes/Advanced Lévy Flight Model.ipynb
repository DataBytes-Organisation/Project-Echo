# Advanced Levy Flight Simulator Animated

The Advanced Levy Flight Simulator Animated is a sophisticated simulation model designed to explore the patterns and behaviors of Levy Flights within a specified environment. This model is particularly useful for studying natural phenomena such as animal foraging patterns, human mobility, and even particle dynamics in various fields of physics and biology.

## Features

- **Environment with Resource Gradient:** The simulator initializes an environment with a gradient to mimic the distribution of resources. This feature allows for the simulation of more realistic scenarios where entities move towards areas of higher concentration.

- **Memory and Adaptation:** Entities in the simulation possess a memory of their environment, which decays over time. They adapt their scale of movement based on the resources encountered, simulating learning and adaptation to the environment.

- **Lévy Flight Movement:** The model employs Lévy flights for movement, characterized by long tails of step lengths. This results in a mix of short, exploratory movements and long, exploratory jumps, a pattern observed in many natural systems.

- **Direction Selection:** Movement direction is influenced by both the environmental gradient and the entity's memory, with a mechanism to smooth transitions and avoid unrealistically sharp turns.

- **Animation:** The simulation is visualized through an animation that displays the path of the entity over time, providing an intuitive understanding of its behavior and the underlying principles of Lévy flights.

## Usage

To run the simulator, instantiate the `AdvancedLevyFlightSimulatorAnimated` class with your desired configuration, and call the `animate` method specifying the number of steps and the animation interval. This will generate an animation that visualizes the path of the entity through its environment, showcasing the dynamics of Lévy flights in response to the simulated conditions.

## Conclusion

The `AdvancedLevyFlightSimulatorAnimated` class offers a powerful tool for researchers, educators, and enthusiasts to explore and understand the complex dynamics of Levy Flights.
```python
class AdvancedLevyFlightSimulatorAnimated:
    def __init__(self, environment_size=(100, 100), memory_decay=0.9, initial_scale=1.0, adaptation_rate=0.1, max_step_length=10, max_turn_angle=np.pi/4):
        self.position = np.array([environment_size[0] / 2, environment_size[1] / 2])
        self.environment = np.zeros(environment_size)
        self.memory = np.zeros(environment_size)
        self.memory_decay = memory_decay
        self.scale = initial_scale
        self.adaptation_rate = adaptation_rate
        self.max_step_length = max_step_length
        self.max_turn_angle = max_turn_angle
        self.current_direction = np.random.rand() * 2 * np.pi
        self.path = [self.position.copy()]
        self._initialize_environment()
        
    def _initialize_environment(self):
        """Create a gradient in the environment to simulate resource distribution."""
        gradient_direction = np.random.rand() * 2 * np.pi
        for i in range(self.environment.shape[0]):
            for j in range(self.environment.shape[1]):
                self.environment[i, j] = i * np.cos(gradient_direction) + j * np.sin(gradient_direction)
        self.environment /= self.environment.max()
        
    def _levy_flight_step(self):
        """Generate a step from a Lévy distribution with a maximum step length."""
        while True:
            step_length = self.scale * (np.power(1 - np.random.rand(), -1.0) - 1)
            if step_length <= self.max_step_length:
                return step_length
        
    def _select_direction(self, combined_gradient):
        """Select direction based on a smoother transition influenced by environmental and memory gradients."""
        if np.linalg.norm(combined_gradient) == 0:
            if len(self.path) > 1:
                prev_direction = np.arctan2(self.path[-1][1] - self.path[-2][1], self.path[-1][0] - self.path[-2][0])
                direction_deviation = (np.random.rand() - 0.5) * np.pi / 4
                return prev_direction + direction_deviation
            else:
                return np.random.rand() * 2 * np.pi
        else:
            return np.arctan2(combined_gradient[1], combined_gradient[0])
    
    def step(self):
        env_gradient = np.gradient(self.environment)
        memory_gradient = np.gradient(self.memory)
        combined_gradient = np.array([
            env_gradient[0][int(self.position[0]), int(self.position[1])] + memory_gradient[0][int(self.position[0]), int(self.position[1])],
            env_gradient[1][int(self.position[0]), int(self.position[1])] + memory_gradient[1][int(self.position[0]), int(self.position[1])]
        ])
        
        new_direction = self._select_direction(combined_gradient)
        
        angle_diff = (new_direction - self.current_direction + np.pi) % (2 * np.pi) - np.pi
        
        if abs(angle_diff) > self.max_turn_angle:
            angle_diff = np.sign(angle_diff) * self.max_turn_angle
        
        self.current_direction = (self.current_direction + angle_diff) % (2 * np.pi)
        
        step_length = min(self._levy_flight_step(), self.max_step_length)
        delta = np.array([np.cos(self.current_direction), np.sin(self.current_direction)]) * step_length
        self.position += delta
        self.position = np.clip(self.position, 0, self.environment.shape[0] - 1)
        
        self.memory *= self.memory_decay
        self.memory[int(self.position[0]), int(self.position[1])] += 1
        self.scale += self.adaptation_rate * (self.environment[int(self.position[0]), int(self.position[1])] - 0.5)
        
        self.path.append(self.position.copy())
        
    def animate(self, steps=100, interval=100):
        fig, ax = plt.subplots(figsize=(8, 8))
        ax.set_xlim(0, self.environment.shape[0])
        ax.set_ylim(0, self.environment.shape[1])
        
        line, = ax.plot([], [], lw=2)
        
        def init():
            line.set_data([], [])
            return line,
        
        def update(frame):
            self.step()
            xdata, ydata = zip(*self.path)
            line.set_data(xdata, ydata)
            return line,
        
        anim = FuncAnimation(fig, update, frames=range(steps), init_func=init, blit=True, interval=interval)
        plt.close()
        return anim
```

## Create and run the animated simulation

```python
simulator = AdvancedLevyFlightSimulatorAnimated(environment_size=(100, 100))
anim = simulator.animate(steps=100, interval=100)
```
